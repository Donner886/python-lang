```python
class FunctionAsDescriptor:
    def __get__(self, obj, cls):
        if obj is None:
            return self
        return lambda : f'Bound to {obj}'
```

```python
class MyClass:
    method = FunctionAsDescriptor()
```

```python
obj = MyClass()
print(obj.method())
```
